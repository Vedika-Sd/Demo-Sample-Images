# NumPy 
Python does not have array, it have lists, but lists are a combination of mixed datatypes and not suitable for mathematical computations; thus, 
for handling large data and mathematical computation, we use NumPy.

## Numpy's Superpower
- Faster than Python lists (C-optimized backend)
- Uses less memory (efficient storage)
- Supports vectorized operations (no explicit loops needed)
- Has built-in mathematical functions

```python
import numpy as np
```

# I. Python List vs. NumPy Array
Let's check speed, so we clearly understand why actually we need NumPy.

```python
import time

size = 100000
list1 = list(range(size))
list2 = list(range(size))

start = time.time()
result = [x+y for x,y in zip(list1,list2)]
end = time.time()

print("Python list addition time:", end - start)
```

```python
import time

size = 100000
arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
result = arr1 +arr2
end = time.time()

print("Python array addition time:", end - start)
```

> Now we see differnce in computaional speed is cruical for handling very large data,
And its proved that numpy arrays are 100x faster than python list

# VII. Vectorization and Broadcasting in Numpy

```python
arr1 = np.array([1,2,3,4,5]) 
result = arr1**2
result            #THis is vectorization, here opertion help fastly using efficient CPu features, bypass slow python list that require loop to do same task
```

```python
arr2 = np.array([11,12,13,14,15])
result = arr2 - 2
result    #This is broadcastination, here 2 is brodcasted to arr2 with size of arr2, doesnt reuire complex opertions
```
